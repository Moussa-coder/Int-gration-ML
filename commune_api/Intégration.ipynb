```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import joblib
```

```python
# Encodage des communes
df = pd.read_csv("donnees_communes_clean.csv")
le = LabelEncoder()
df["Commune_enc"] = le.fit_transform(df["Commune"])
df
```

```python
# Création des variables d'entrée (X) et des cibles (y)
X = df[["Année", "Commune_enc"]]
y_recettes = df["Recettes (M€)"]
y_depenses = df["Dépenses (M€)"]
X
```

```python
# Séparation des données en train/test
X_train, X_test, y_train_rec, y_test_rec = train_test_split(X, y_recettes, test_size=0.2)
X_train, X_test, y_train_dep, y_test_dep = train_test_split(X, y_depenses, test_size=0.2)
print("=== X_train ===")
print(X_train.head())
print("\n=== y_train_rec (Recettes) ===")
print(y_train_rec.head())

print("\n=== y_train_dep (Dépenses) ===")
print(y_train_dep.head())
```

```python
print("\n=== X_test ===")
print(X_test.head())

print("\n=== y_test_rec (Recettes Test) ===")
print(y_test_rec.head())

print("\n=== y_test_dep (Dépenses Test) ===")
print(y_test_dep.head())
```

```python
# Entraînement des modèles RandomForest
model_rec = RandomForestRegressor().fit(X_train, y_train_rec)
model_dep = RandomForestRegressor().fit(X_train, y_train_dep)
```

```python
# Sauvegarde des modèles entraînés et de l'encodeur
joblib.dump(model_rec, "model_recettes.pkl")
joblib.dump(model_dep, "model_depenses.pkl")
joblib.dump(le, "label_encoder.pkl")
```
